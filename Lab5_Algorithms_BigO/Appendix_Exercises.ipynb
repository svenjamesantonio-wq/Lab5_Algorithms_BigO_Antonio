```python
# 1. Emergency Dispatch Record Search
# Student-Specific Inputs
LAST_NAME = "Antonio"
STUDENT_ID = "TUPM-26-1833"
FAVORITE_ARTIST = "Radiohead"

# Derived Inputs Calculation
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(char) for char in STUDENT_ID if char.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)
ARTIST_SUM = sum(ord(char.upper()) - ord('A') + 1 for char in FAVORITE_ARTIST if char.isalpha())

print("=== STUDENT-SPECIFIC INPUTS & DERIVED VALUES ===")
print(f"  * Last Name: {LAST_NAME} (Length: {NAME_LENGTH})")
print(f"  * Student ID: {STUDENT_ID} (ID Sum: {ID_SUM}, Seed Digit: {SEED_DIGIT})")
print(f"  * Favorite Artist: {FAVORITE_ARTIST} (Length: {ARTIST_LENGTH}, Artist Sum: {ARTIST_SUM})")

# 1. Generate Original Dataset
original_dataset = sorted([ID_SUM + (i * (SEED_DIGIT + 2)) for i in range(15)])
print("\n=== 1. ORIGINAL DATASET ===")
print(original_dataset)

# 2. Generate Rotated Dataset
rotation_point = SEED_DIGIT % len(original_dataset)
rotated_dataset = original_dataset[rotation_point:] + original_dataset[:rotation_point]
print("\n=== 2. ROTATED DATASET ===")
print(f"  * Rotation Pivot Index: {rotation_point}")
print(rotated_dataset)

# 3. Generate Target Incident
target_index = (ID_SUM + ARTIST_SUM) % len(original_dataset)
target_incident = original_dataset[target_index]
print("\n=== 3. TARGET INCIDENT ===")
print(f"  * Target Incident Value: {target_incident}")

# --- SEARCH ALGORITHMS ---

def search_rotated_binary(arr, target):
    left, right = 0, len(arr) - 1
    comparisons = 0
    while left <= right:
        comparisons += 1
        mid = (left + right) // 2
        if arr[mid] == target:
            return mid, comparisons
        if arr[left] <= arr[mid]:
            comparisons += 1
            if arr[left] <= target < arr[mid]:
                right = mid - 1
            else:
                left = mid + 1
        else:
            comparisons += 1
            if arr[mid] < target <= arr[right]:
                left = mid + 1
            else:
                right = mid - 1
    return -1, comparisons

def search_linear(arr, target):
    comparisons = 0
    for idx, val in enumerate(arr):
        comparisons += 1
        if val == target:
            return idx, comparisons
    return -1, comparisons

rot_index, rot_comparisons = search_rotated_binary(rotated_dataset, target_incident)
lin_index, lin_comparisons = search_linear(rotated_dataset, target_incident)

# 4. Search Position / Report if found
print("\n=== 4. SEARCH POSITION & RESULT ===")
if rot_index != -1:
    print(f"  * Target incident {target_incident} successfully located at Index: {rot_index}")
else:
    print(f"  * Target incident {target_incident} is not found.")

# 5. Comparison Counts & Linear Search Comparison
print("\n=== 5. COMPARISON COUNTS ===")
print(f"  * Rotated Binary Search Comparisons: {rot_comparisons}")
print(f"  * Linear Search comparison count: {lin_comparisons} (Found at Index: {lin_index})")

# 6. Time and Space Complexity
print("\n=== 6. TIME AND SPACE COMPLEXITY ===")
print("  * Rotated Binary Search:")
print("    - Time Complexity: O(log n)")
print("    - Space Complexity: O(1)")
print("  * Linear Search:")
print("    - Time Complexity: O(n)")
print("    - Space Complexity: O(1)")

# --- FINAL OUTPUT SUMMARY BLOCK ---
print("\n" + "="*40)
print("FINAL OUTPUT")
print("="*40)
print(f"  * Original Dataset: {original_dataset}")
print(f"  * Rotated Dataset: {rotated_dataset}")
print(f"  * Target Incident: {target_incident}")
print(f"  * Rotated Binary Search Result: Found at Index {rot_index} ({rot_comparisons} comparisons)")
print(f"  * Linear Search Result: Found at Index {lin_index} ({lin_comparisons} comparisons)")
print("  * Status: Successfully completed and verified.")
```

```python
# 2. University Enrollment Records
# Student-Specific Inputs
LAST_NAME = "Antonio"
STUDENT_ID = "TUPM-26-1833"
FAVORITE_ARTIST = "Radiohead"

# Derived Inputs Calculation
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(char) for char in STUDENT_ID if char.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

print("=== STUDENT-SPECIFIC INPUTS & DERIVED VALUES ===")
print(f"  * Last Name: {LAST_NAME} (Length: {NAME_LENGTH})")
print(f"  * Student ID: {STUDENT_ID} (ID Sum: {ID_SUM}, Seed Digit: {SEED_DIGIT})")
print(f"  * Favorite Artist: {FAVORITE_ARTIST} (Length: {ARTIST_LENGTH})")

# 1. Generate the two sorted enrollment datasets
dept_a = sorted([ID_SUM + (i * (SEED_DIGIT + 2)) for i in range(7)])
dept_b = sorted([(ARTIST_LENGTH * 3) + (i * (NAME_LENGTH + 1)) for i in range(7)])

print("\n=== 1. DEPARTMENT DATASETS ===")
print(f"  * Department A dataset: {dept_a}")
print(f"  * Department B dataset: {dept_b}")

# 2. Consolidate them into one ordered collection
def merge_sorted_datasets(arr1, arr2):
    merged = []
    i = 0
    j = 0
    comparisons = 0
    while i < len(arr1) and j < len(arr2):
        comparisons += 1
        if arr1[i] <= arr2[j]:
            merged.append(arr1[i])
            i += 1
        else:
            merged.append(arr2[j])
            j += 1
    while i < len(arr1):
        merged.append(arr1[i])
        i += 1
    while j < len(arr2):
        merged.append(arr2[j])
        j += 1
    return merged, comparisons

consolidated_dataset, comparison_count = merge_sorted_datasets(dept_a, dept_b)

print("\n=== 2. CONSOLIDATED DATASET ===")
print(f"  * Consolidated dataset: {consolidated_dataset}")

# Record counts calculation
records_before = len(dept_a) + len(dept_b)
records_after = len(consolidated_dataset)

print("\n=== 3. NUMBER OF RECORDS BEFORE AND AFTER CONSOLIDATION ===")
print(f"  * Before consolidation: {records_before} records (Dept A: {len(dept_a)}, Dept B: {len(dept_b)})")
print(f"  * After consolidation: {records_after} records")

# 4. Count comparisons performed
print("\n=== 4. COMPARISON COUNT ===")
print(f"  * Comparison count: {comparison_count}")

# 5. Verify final order
is_sorted = all(consolidated_dataset[k] <= consolidated_dataset[k+1] for k in range(len(consolidated_dataset) - 1))
ordering_status = "Passed (Ascending Order Verified)" if is_sorted else "Failed"

print("\n=== 5. ORDERING VERIFICATION ===")
print(f"  * Ordering verification: {ordering_status}")

# 6. Time and space complexity
print("\n=== 6. TIME AND SPACE COMPLEXITY ===")
print("  * Time and space complexity: Time O(n + m), Space O(n + m)")

# --- FINAL OUTPUT SUMMARY BLOCK (Top-to-Bottom Checklist Order) ---
print("\n" + "="*40)
print("FINAL OUTPUT")
print("="*40)
print(f"  1. Department A dataset: {dept_a}")
print(f"  2. Department B dataset: {dept_b}")
print(f"  3. Consolidated dataset: {consolidated_dataset}")
print(f"  4. Number of records before and after consolidation: Before = {records_before}, After = {records_after}")
print(f"  5. Comparison count: {comparison_count}")
print(f"  6. Ordering verification: {ordering_status}")
print("  7. Time and space complexity: Time O(n + m), Space O(n + m)")
print(f"  * Final Output: Successfully consolidated {records_before} records into {records_after} sorted records.")
```

```python
# 3. E-Commerce Order Processing
import time
import random

# Student-Specific Inputs
LAST_NAME = "Antonio"
STUDENT_ID = "TUPM-26-1833"
FAVORITE_ARTIST = "Radiohead"

# Derived Inputs Calculation
SEED_DIGIT = int(STUDENT_ID[-1])
ID_SUM = sum(int(char) for char in STUDENT_ID if char.isdigit())
NAME_LENGTH = len(LAST_NAME)
ARTIST_LENGTH = len(FAVORITE_ARTIST)

print("=== STUDENT-SPECIFIC INPUTS & DERIVED VALUES ===")
print(f"  * Last Name: {LAST_NAME} (Length: {NAME_LENGTH})")
print(f"  * Student ID: {STUDENT_ID} (ID Sum: {ID_SUM}, Seed Digit: {SEED_DIGIT})")
print(f"  * Favorite Artist: {FAVORITE_ARTIST} (Length: {ARTIST_LENGTH})")

# Sorting Algorithms Implementation
def bubble_sort(arr):
    data = arr.copy()
    comparisons = 0
    movements = 0
    start_time = time.perf_counter()
    n = len(data)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            comparisons += 1
            if data[j] > data[j + 1]:
                data[j], data[j + 1] = data[j + 1], data[j]
                movements += 3
                swapped = True
        if not swapped:
            break
    end_time = time.perf_counter()
    return data, comparisons, movements, (end_time - start_time) * 1000

def selection_sort(arr):
    data = arr.copy()
    comparisons = 0
    movements = 0
    start_time = time.perf_counter()
    n = len(data)
    for i in range(n):
        min_idx = i
        for j in range(i + 1, n):
            comparisons += 1
            if data[j] < data[min_idx]:
                min_idx = j
        if min_idx != i:
            data[i], data[min_idx] = data[min_idx], data[i]
            movements += 3
    end_time = time.perf_counter()
    return data, comparisons, movements, (end_time - start_time) * 1000

def insertion_sort(arr):
    data = arr.copy()
    comparisons = 0
    movements = 0
    start_time = time.perf_counter()
    for i in range(1, len(data)):
        key = data[i]
        movements += 1
        j = i - 1
        while j >= 0:
            comparisons += 1
            if data[j] > key:
                data[j + 1] = data[j]
                movements += 1
                j -= 1
            else:
                break
        data[j + 1] = key
        movements += 1
    end_time = time.perf_counter()
    return data, comparisons, movements, (end_time - start_time) * 1000

# Base Data Generation
base_orders = sorted([ID_SUM + (i * (SEED_DIGIT + 3)) for i in range(10)])
sorted_inputs = base_orders.copy()
reverse_inputs = sorted_inputs[::-1]
random_inputs = base_orders.copy()
random.seed(ID_SUM)
random.shuffle(random_inputs)

input_conditions = {
    "Sorted": sorted_inputs,
    "Reverse": reverse_inputs,
    "Random": random_inputs
}

# Execute all algorithms
results = {}
for cond_name, dataset in input_conditions.items():
    b_sorted, b_comp, b_mov, b_time = bubble_sort(dataset)
    s_sorted, s_comp, s_mov, s_time = selection_sort(dataset)
    i_sorted, i_comp, i_mov, i_time = insertion_sort(dataset)
    
    results[cond_name] = {
        "Bubble": (b_sorted, b_comp, b_mov, b_time, all(b_sorted[k] <= b_sorted[k+1] for k in range(len(b_sorted)-1))),
        "Selection": (s_sorted, s_comp, s_mov, s_time, all(s_sorted[k] <= s_sorted[k+1] for k in range(len(s_sorted)-1))),
        "Insertion": (i_sorted, i_comp, i_mov, i_time, all(i_sorted[k] <= i_sorted[k+1] for k in range(len(i_sorted)-1)))
    }

# --- ASSESSMENT DATA & FINAL OUTPUT (Strictly Separated Top-to-Bottom Checklist) ---
print("\n" + "="*50)
print("FINAL OUTPUT")
print("="*50)

print(f"1. Generated order dataset: {base_orders}")

print("2. Sorted, reverse, and random inputs:")
for cond, data in input_conditions.items():
    print(f"   - {cond} Input: {data}")

print("3. Comparison counts:")
for cond, algs in results.items():
    print(f"   - {cond} Condition -> Bubble: {algs['Bubble'][1]}, Selection: {algs['Selection'][1]}, Insertion: {algs['Insertion'][1]}")

print("4. Movement counts:")
for cond, algs in results.items():
    print(f"   - {cond} Condition -> Bubble: {algs['Bubble'][2]}, Selection: {algs['Selection'][2]}, Insertion: {algs['Insertion'][2]}")

print("5. Execution times:")
for cond, algs in results.items():
    print(f"   - {cond} Condition -> Bubble: {algs['Bubble'][3]:.4f}ms, Selection: {algs['Selection'][3]:.4f}ms, Insertion: {algs['Insertion'][3]:.4f}ms")

print("6. Sorting verification:")
for cond, algs in results.items():
    print(f"   - {cond} Condition Verified -> Bubble: {algs['Bubble'][4]}, Selection: {algs['Selection'][4]}, Insertion: {algs['Insertion'][4]}")

print("7. Big O analysis:")
print("   - Bubble Sort: O(n^2) worst/average, O(n) best")
print("   - Selection Sort: O(n^2) across all cases")
print("   - Insertion Sort: O(n^2) worst/average, O(n) best")

print("8. Best-, average-, and worst-case observations:")
print("   - Insertion and Bubble Sort adaptively perform in O(n) time on pre-sorted inputs with early exits or zero shifts.")
print("   - Selection Sort performs uniform O(n^2) scans regardless of initial order arrangement.")

print("9. Final Output: E-commerce order processing sorting performance successfully simulated, tracked, and verified across all input conditions.")
```
